## Nhận thấy cột project_id null khá nhiều, cột notes cũng null nhiều nhưng không phải là vấn đề

```python
df.head(10)
```

# 1. Xử lý dữ liệu null

```python
#Thực hiện lọc những project_id bất thường
df.loc[~df['project_id'].str.contains('PRJ', na=False), 'project_id'].unique()
```

## Vậy ngoại trừ những id hợp lệ thì chỉ có null và '-', không phải đơn hàng nào cũng có mã dự án nên null không được coi là bất thường

```python
#Thực hiện replace
df['project_id'] = df['project_id'].replace('-', np.nan)
```

```python
#Kiểm tra lại
df.loc[~df['project_id'].str.contains('PRJ', na=False), 'project_id'].unique()
```

# 2. Xử lý dữ liệu thiếu nhất quán

```python
df.loc[df['quantity'].str.contains('chiếc|bộ|cái', na = False), 'quantity'].unique()
```

```python
# Sau khi nhận thấy những dòng có chứ ký tự có quy luật số - ' ' - ký tự thì chọn cách tạo hàm split để lấy ra số 
```

```python
#chuyển type qua str 
df['quantity'] = df['quantity'].astype(str)
```

```python
# Tạo hàm tách chuỗi
def ham_split(x):
    s1 = x.split(' ')
    if len(s1)==2:
        return s1[0]
    else:
        return s1[0]
```

```python
#apply vào cột quantity
df['quantity'] = df['quantity'].apply(ham_split)
```

```python
#kiểm tra
df.loc[df['quantity'].str.contains('chiếc|bộ|cái', na = False), 'quantity'].unique()
```

```python
#Chuyển lại kiểu int
df['quantity'] = df['quantity'].astype(int)
```

```python
#Chuyển dữ liệu thiếu nhất quán cột discount
mask = df['discount'].str.contains('%', na=False)

df.loc[mask, 'discount'] = (df.loc[mask, 'discount'].str.replace('%', '', regex=False).astype(float)/ 100)
```

```python
#chuyển type qua str 
df['unit_price'] = df['unit_price'].astype(str)
```

```python
#chuyển đổi dữ liệu thiếu nhất quán xóa kỹ hiệu đ
df['unit_price'] = df['unit_price'].apply(ham_split)
```

```python
#Xóa dấu , chuyển lại về kiểu float
df['unit_price'] = df['unit_price'].str.replace(',', '').astype(float)
```

## 3. Xử lý dữ liệu dạng text

```python
#Loại bỏ khoảng trắng thừa
df['sales_channel'] = df['sales_channel'].str.strip()
```

```python
#Loại bỏ khoảng trắng thừ chuyển text về title
df['payment_method'] = df['payment_method'].str.strip().str.title()
```

```python
#Loại bỏ khoảng trắng thừ chuyển text về title
df['order_status'] = df['order_status'].str.strip().str.title()
```

```python
#Chuyển định dạng ngày tháng về 1 kiểu đồng nhất
df['order_date'] = pd.to_datetime(df['order_date'],format='mixed',errors='coerce').dt.date
df['delivery_date'] = pd.to_datetime(df['delivery_date'],format='mixed',errors='coerce').dt.date
```

```python
df.head(10)
```

```python
df['order_status'] = df['order_status'].replace('Đã Giao', 'Hoàn Thành')
```

## 4. Xử lý dữ liệu trùng lặp

```python
#hiển thị dữ liệu trùng lặp
df[df.duplicated(keep = False)].sort_values('order_id')
```

```python
#Xử lý dữ liệu trùng lặp giữ lại bản ghi đầu tiên
df = df.drop_duplicates(keep = 'first')
```

## 5. Ghi đè dữ liệu đã làm sạch vào file cũ

```python
excel_path = 'dataset/IT_equipment_business.xlsx'
sheet_target = 'orders'

# Ghi đè biến df đã làm sạch vào chính sheet 'orders'
with pd.ExcelWriter(
    excel_path,
    engine='openpyxl',
    mode='a',                   # Mở file hiện có để chỉnh sửa
    if_sheet_exists='replace'   # Thay thế/ghi đè sheet cũ, giữ nguyên các sheet khác
) as writer:
    df.to_excel(writer, sheet_name=sheet_target, index=False)

print(f"Đã ghi đè thành công dữ liệu sạch vào sheet '{sheet_target}'!")
```
